# identify_locations.ipynb
Identify FACTS locations that contain a tide gaugue or city and save these locations to `data_locations/location.lst`.

```python
import d25a
import datetime
from pathlib import Path
```

```python
# Get start datetime
start_dt = datetime.datetime.now()
print(start_dt.strftime('%Y-%m-%d %H:%M:%S'))
```

```python
# Print package versions
print(d25a.get_watermark())
```

```python
# Get coastal locations of interest
coastal_loc_df = d25a.get_coastal_loc_df()
coastal_loc_df
```

```python
# Save locations to file
out_dir = Path("data_locations")
out_dir.mkdir(parents=True, exist_ok=True)
out_fn = out_dir / "location.lst"
coastal_loc_df.to_csv(out_fn, sep="\t", index=False, header=False, float_format="%.2f")
print(f"Saved to {out_fn} ({out_fn.stat().st_size / 1024 / 1024:.2f} MB)")
```

```python
# Check that formatting of common lines is same as original FACTS location.lst
orig_fn = d25a.DOWNLOADED_DIR / 'location.lst'
with open(orig_fn, 'r') as f:
    orig_lines = f.readlines()
    orig_set = set(orig_lines)
with open(out_fn, 'r') as f:
    out_lines = f.readlines()
    out_set = set(out_lines)
print(f"Shared lines: {len(orig_set & out_set)}")
print(f"Lines only in original location.lst: {len(orig_set - out_set)}")
print(f"Lines only in new location.lst: {len(out_set - orig_set)}")
# Checks
assert len(out_set) == len(out_lines), "New location.lst contains duplicate lines."
assert not out_set - orig_set, "New location.lst contains lines that do not match original."
assert len(out_lines) == len(coastal_loc_df), "Number of output lines does not match number of locations."
```

```python
# Get end datetime
end_dt = datetime.datetime.now()
# Calculate run timedelta
run_td = end_dt - start_dt
# Print timing information
print(f"Start:     {start_dt.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"End:       {end_dt.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"Wall time: {run_td.total_seconds():.0f} s")
```
